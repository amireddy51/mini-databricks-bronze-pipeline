```python
# Step 1: Define the source file path
source_path = "/Volumes/workspace/customer_data_pipeline/landing/customers.csv"

# Step 2: Read the CSV into a Spark DataFrame
df_bronze = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(source_path)
)

# Step 3: Display the raw customer records
display(df_bronze)
```

```python
# Bronze Layer - Save raw customer data as a Delta table

bronze_table = "workspace.customer_data_pipeline.bronze_customers"

(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(bronze_table)
)

print("Bronze Delta table created successfully!")
```

```python
# Bronze Layer - Basic data validation

bronze_table = "workspace.customer_data_pipeline.bronze_customers"

# Read the saved Delta table
df_check = spark.table(bronze_table)

# Check total records
record_count = df_check.count()

print(f"Total Bronze records: {record_count}")

# Validate the expected number of records
assert record_count == 5, "Bronze record count does not match!"

print("Bronze record count validation PASSED")
```
